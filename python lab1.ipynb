```python
import numpy as np
import matplotlib.pyplot as plt
import labx as mx

student_id = 2600562

# x values
x = np.arange(-10, 10, 0.001)

# y values
y = mx.lab_1_problem_1(x, student_id)

# پیدا کردن نقاطی که علامت عوض می‌شود (تقاطع با y=0)
sign_changes = np.where(np.diff(np.sign(y)))[0]

# گرفتن دو نقطه اول
x_intersections = x[sign_changes[:2]]
y_intersections = y[sign_changes[:2]]

# رسم نمودار
plt.plot(x, y, label="f(x)")

# مشخص کردن نقاط برخورد
plt.scatter(x_intersections, y_intersections, s=80, label="Intersections")

# نوشتن مختصات کنار نقاط
for xi, yi in zip(x_intersections, y_intersections):
    plt.text(xi, yi, f"({xi:.3f}, {yi:.3f})")

plt.title("Function f(x) for student 2600562")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# پیدا کردن نقاطی که تابع از y=1 عبور می‌کند
crossings = np.where(np.diff(np.sign(y - 1)))[0]

# x های تقریبی نقاط برخورد
x_cross = x[crossings]

# ---- Graph for first intersection ----
plt.figure(figsize=(7,5))
plt.plot(x, y)
plt.xlim(x_cross[0]-0.5, x_cross[0]+0.5)   # zoom around first intersection
plt.ylim(0, 2)
plt.axhline(y=1, color='r', linestyle='--')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Point 1 - Zoomed view')
plt.grid(True)
plt.show()

# ---- Graph for second intersection ----
plt.figure(figsize=(7,5))
plt.plot(x, y)
plt.xlim(x_cross[1]-0.5, x_cross[1]+0.5)   # zoom around second intersection
plt.ylim(0, 2)
plt.axhline(y=1, color='r', linestyle='--')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Point 2 - Zoomed view')
plt.grid(True)
plt.show()
```
